Аналіз датасету Individual Household Electric Power Consumption.

Звантажити та відкрити датасет. Здійснити data cleaning.

```python
import pandas as pd
import numpy as np
import timeit
import warnings
warnings.filterwarnings('ignore') 

def load_and_clean_power_data(filepath="household_power_consumption.txt"):
    print("Починаємо читати файл")
    df = pd.read_csv(filepath, sep=';', na_values=['?'], low_memory=False)
    print("Файл зчитано. Видаляємо пропуски")
    df = df.dropna()
    print("Конвертуємо колонки Date та Time у справжній формат часу")
    df['Datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d/%m/%Y %H:%M:%S')
    print("Перетворюємо текстові цифри на числа з плаваючою комою")
    numeric_cols = ['Global_active_power', 'Global_reactive_power', 'Voltage', 
                    'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']
    for col in numeric_cols:
        df[col] = df[col].astype(float)   
    print(f"Очищення завершено. Залишилося чистих рядків: {len(df)}")
    return df

start_time = timeit.default_timer()
df_power = load_and_clean_power_data()
execution_time = timeit.default_timer() - start_time
print(f"Час виконання завантаження та очищення: {execution_time:.2f} секунд")
display(df_power.head())
```

Обрати всі записи, у яких загальна активна споживана потужність (Global_active_power) перевищує 5 кВт.

```python
def filter_high_power(dataframe):
    return dataframe[dataframe['Global_active_power'] > 5.0]
start_time = timeit.default_timer()
high_power_df = filter_high_power(df_power)
exec_time = timeit.default_timer() - start_time
print(f"Результати")
print(f"Знайдено записів (> 5 кВт): {len(high_power_df)}")
print(f"Час виконання: {exec_time:.4f} секунд")
display(high_power_df.head())
```

Обрати всі записи, у яких сила струму лежить в межах 19-20 А, для них виявити ті, у яких пральна машина та холодильник (Sub_metering_2) споживають більше, ніж бойлер та кондиціонер (Sub_metering_3).

```python
def filter_appliances(dataframe):
    filtered = dataframe[(dataframe['Global_intensity'] >= 19.0) & (dataframe['Global_intensity'] <= 20.0)]
    result = filtered[filtered['Sub_metering_2'] > filtered['Sub_metering_3']]
    return result
    
start_time = timeit.default_timer()
appliances_df = filter_appliances(df_power)
exec_time = timeit.default_timer() - start_time

print(f"Результати")
print(f"Знайдено записів (19-20 А та Група 2 > Група 3): {len(appliances_df)}")
print(f"Час виконання: {exec_time:.4f} секунд")
display(appliances_df.head())
```

Обрати випадковим чином 500000 записів (без повторів елементів вибірки), для них обчислити середні величини усіх 3-х груп споживання електричної енергії.

```python
def random_sample_means(dataframe):
    sample_df = dataframe.sample(n=500000, replace=False)
    means = sample_df[['Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']].mean()
    return means

start_time = timeit.default_timer()
means_result = random_sample_means(df_power)
exec_time = timeit.default_timer() - start_time
print(f"Результати")
print(f"Час виконання: {exec_time:.4f} секунд")
print("Середні значення споживання (Ватт-годин):")
print(means_result)
```

Обрати ті записи, які після 18-00 споживають понад 6 кВт за хвилину в середньому, серед відібраних визначити ті, у яких основне споживання електроенергії у вказаний проміжок часу припадає на пральну машину, сушарку, холодильник та освітлення (група 2 є найбільшою), а потім обрати кожен третій результат із першої половини та кожен четвертий результат із другої половини.

```python
def complex_evening_filter(dataframe):
    evening_df = dataframe[dataframe['Datetime'].dt.hour >= 18]
    high_power = evening_df[evening_df['Global_active_power'] > 6.0]
    group2_max = high_power[
        (high_power['Sub_metering_2'] > high_power['Sub_metering_1']) &
        (high_power['Sub_metering_2'] > high_power['Sub_metering_3'])
    ]

    half_index = len(group2_max) // 2
    first_half = group2_max.iloc[:half_index]
    second_half = group2_max.iloc[half_index:]
    first_half_res = first_half.iloc[::3]
    second_half_res = second_half.iloc[::4]
    final_result = pd.concat([first_half_res, second_half_res]) 
    return final_result

start_time = timeit.default_timer()
complex_df = complex_evening_filter(df_power)
exec_time = timeit.default_timer() - start_time

print(f"Результати")
print(f"Знайдено фінальних записів: {len(complex_df)}")
print(f"Час виконання: {exec_time:.4f} секунд")
display(complex_df.head())
```

Пронормувати та стандартизувати вибраний датасет.

```python
def normalize_and_standardize(dataframe):
    df_math = dataframe[['Global_active_power', 'Global_intensity']].copy()
    min_val = df_math['Global_active_power'].min()
    max_val = df_math['Global_active_power'].max()
    df_math['Active_power_NORMALIZED'] = (df_math['Global_active_power'] - min_val) / (max_val - min_val)
    mean_val = df_math['Global_intensity'].mean()
    std_val = df_math['Global_intensity'].std()
    df_math['Intensity_STANDARDIZED'] = (df_math['Global_intensity'] - mean_val) / std_val
    return df_math
    
df_scaled = normalize_and_standardize(df_power)
print("Результат нормалізації та стандартизації")
display(df_scaled.head())
```

Підрахувати коефіцієнт Пірсона та Спірмена для двох integer/real атрибутів.

```python
def calculate_correlations(dataframe, col1, col2):
    pearson_corr = dataframe[col1].corr(dataframe[col2], method='pearson')
    spearman_corr = dataframe[col1].corr(dataframe[col2], method='spearman')
    print(f"Кореляція між '{col1}' та '{col2}'")
    print(f"Коефіцієнт Пірсона:  {pearson_corr:.4f}")
    print(f"Коефіцієнт Спірмена: {spearman_corr:.4f}")
    if pearson_corr > 0.9:
        print("Між цими показниками існує дуже сильна пряма залежність")

calculate_correlations(df_power, 'Global_active_power', 'Global_intensity')
```

Провести One Hot Encoding категоріального атрибута

```python
def perform_one_hot_encoding(dataframe):
    print("1. Створюємо категоріальний атрибут")
    df_sample = dataframe.head(10000).copy()
    df_sample['Day_of_Week'] = df_sample['Datetime'].dt.day_name()
    print("2. Проводимо One Hot Encoding")
    df_encoded = pd.get_dummies(df_sample, columns=['Day_of_Week'])
    return df_encoded

df_ohe = perform_one_hot_encoding(df_power)
print("\nРезультат One Hot Encoding")
ohe_columns = [col for col in df_ohe.columns if 'Day_of_Week' in col]
display(df_ohe[ohe_columns].head())
```
